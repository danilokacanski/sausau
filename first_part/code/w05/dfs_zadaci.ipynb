## Zadatak 1: Da li postoji put?

Dat je usmeren graf. Napisati funkciju koja pomoću DFS algoritma proverava da li postoji put
od čvora ’A’ do čvora ’F’.

```python
def dfs(graph, node, goal, visited):
    if node == goal:    # Ako smo dosli do cilja, vracamo True
        return True

    visited.add(node)   # Dodajemo trenutni cvor u skup posecenih cvorova, da ne bismo ponovo posecivali isti cvor

    for neighbor in graph[node]:    # Prolazimo kroz sve susede trenutnog cvora
        if neighbor not in visited:     # Ako sused nije posecen, nastavljamo pretragu od tog suseda
            if dfs(graph, neighbor, goal, visited):     # Ako je pretraga od suseda vratila True, to znaci da smo nasli cilj, pa i mi vracamo True
                return True

    return False

def dfs2(graph, node, goal, visited):
    if node == goal:    # Ako smo dosli do cilja, vracamo True
        return True

    if node in visited:   # Ako je trenutni cvor vec posecen, vracamo False da ne bismo ponovo posecivali isti cvor
        return False
    visited.add(node)   # Dodajemo trenutni cvor u skup posecenih cvorova, da ne bismo ponovo posecivali isti cvor

    for neighbor in graph[node]:    # Prolazimo kroz sve susede trenutnog cvora
        if dfs(graph, neighbor, goal, visited):     # Ako je pretraga od suseda vratila True, to znaci da smo nasli cilj, pa i mi vracamo True
            return True

    return False


def has_path_dfs(graph, start, goal):
    visited = set()    
    return dfs(graph, start, goal, visited)
```

```python
graph = {
    'A': ['B'],
    'B': ['C', 'D'],
    'C': [],
    'D': ['E'],
    'E': ['F'],
    'F': []
}
```

```python
print(has_path_dfs(graph, 'A', 'F'))
```

## Zadatak 2: Broj povezanih komponenti

Dat je neusmeren graf. Odrediti broj povezanih komponenti korišćenjem DFS algoritma. Povezana komponenta je maksimalan skup čvorova između kojih postoji put.

```python
def dfs(graph, node, visited):
    visited.add(node)   # Necemo isti cvor da posecujemo vise puta

    for neighbor in graph[node]:    
        if neighbor not in visited: # Za svaki sused koji nije posecen, radimo idemo dublje
            dfs(graph, neighbor, visited)


def count_components_dfs(graph):
    visited = set()
    components = 0

    for node in graph:  # Ako graf nije povezan, onda ce biti vise komponenti
        if node not in visited:     # Ako smo obisli iz nekog cvora vec, on pripada toj komponenti
            components += 1         # Ako nismo, uvecavamo broj komponenti
            dfs(graph, node, visited)    # Oblizimo sve sto mozemo iz tog cvora

    return components
```

```python
graph = {
    'A': ['B'],
    'B': ['A'],
    'C': ['D'],
    'D': ['C'],
    'E': []
}
```

```python
print(count_components_dfs(graph))
```

## Zadatak 3: Detekcija ciklusa u usmerenom grafu

Dat je usmeren graf. Proveriti pomoću DFS algoritma da li graf sadrži ciklus.

```python
def dfs_cycle(graph, node, visited, recursion_stack):
    visited.add(node)   # Pratimo generalno posecene
    recursion_stack.add(node)   # Pratimo posecene u trenutnoj putanji, da bismo detektovali ciklus

    for neighbor in graph[node]:
        if neighbor not in visited:    # Ako nismo posecivali tog suseda, nastavljamo pretragu od tog suseda, jer nema ciklusa
            if dfs_cycle(graph, neighbor, visited, recursion_stack):
                return True
        elif neighbor in recursion_stack:   # Ako smo posetili tog suseda, i nalazi se u trenutnoj putanji to znaci
                                            # da je ciklus, i odavde se otpetljava rekurzija, vracajuci True za sve do 
                                            # pocetka ciklusa, i na kraju ide u has_cycle_dfs, koja vraca True, 
                                            # jer smo nasli ciklus
            return True

    recursion_stack.remove(node)
    return False


def has_cycle_dfs(graph):
    visited = set()
    recursion_stack = set()

    for node in graph:  # Prolazimo kroz sve cvorove, ne znamo da li je graf povezan, pa mozda ima vise komponenti
        if node not in visited:     # Samo ako nije u posecenim radimo dfs_cycle, jer ako jeste, vec smo ispitali taj ciklus
            if dfs_cycle(graph, node, visited, recursion_stack):
                return True

    return False
```

```python
graph = {
    'A': ['B'],
    'B': ['C'],
    'C': ['D'],
    'D': ['B'],
    'E': []
}
```

```python
print(has_cycle_dfs(graph))
```
