# BFS Zadaci

## Zadatak 1: Najkraci put u netezinskom grafu

Dat je neusmeren graf. Napisati funkciju koja pomoću BFS algoritma pronalazi najkraći put od
čvora *A* do čvora *F*.

```python
from collections import deque

def bfs_shortest_path(graph, start, goal):
    queue = deque()
    queue.append(start)
    visited = set()
    visited.add(start)
    parent = {start: None}  # Pamti iz kog cvora smo prvi put dosli do nekog drugog cvora

    while queue:
        node = queue.popleft()

        if node == goal:    # Kada se pronađe cilj, rekonstruisati put koristeći parent dict
            path = []
            while node is not None:     # None ce biti za start node, jer nema roditelja
                path.append(node)       # U putanju se dodaje prvo cilje, pa njegov roditelj...
                node = parent[node]     # Ovako se krece do start node
            return path[::-1]           # Ocemo na kraju da obrnemo putnju

        for neighbor in graph[node]:    # Dodeljuje parent[neighbor] = node samo ako taj neighbor nije vec posecen, 
                                        # jer ako jeste onda znamo da smo do njega dosli iz nekog drugog cvora, 
                                        # a ne iz ovog trenutnog
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = node
                queue.append(neighbor)

    return None
```

Ovo garantuje da je putanja minimalna, jer BFS redom posecuje najblize cvorove, tj. svoje susede, uz modifikaciju ovde, da pratimo koji su poseceni. 

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'F'],
    'E': ['C', 'F'],
    'F': ['D', 'E']
}
```

```python
print(bfs_shortest_path(graph, 'A', 'F'))
```

## Zadatak 2: Rastojanje od početnog čvora

Dat je graf. Napisati funkciju koja pomoću BFS algoritma računa minimalan broj grana od
početnog čvora *A* do svih ostalih čvorova.

```python
from collections import deque

def bfs_distances(graph, start):
    queue = deque()
    queue.append(start)
    visited = set()
    visited.add(start)
    distance = {start: 0}   # Negde moramo cuvati udaljenost od start

    while queue:
        node = queue.popleft()

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                distance[neighbor] = distance[node] + 1  # Svaki neighbor je udaljen za 1 od njegovog node-a
                queue.append(neighbor)

    return distance
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
```

```python
print(bfs_distances(graph, 'A'))
```

## Zadatak 3: Provera bipartitnosti

Dat je neusmeren graf. Proveriti pomoću BFS algoritma da li je graf bipartitan.
Graf je bipartitan ako se njegovi čvorovi mogu obojiti sa dve boje tako da susedni čvorovi
nemaju istu boju.

```python
from collections import deque

def is_bipartite_bfs(graph):
    color = {}  # Cuvamo boje svakog cvora kao 0 ili 1

    for start in graph:     # Moramo da prodjemo kroz sve cvorove, jer graf moze biti nepovezan, nije dovoljan samo jedan BFS od nekog startnog cvora
        if start not in color:  # Ako cvor nije obojen, radimo BFS od njega
            queue = deque()
            queue.append(start)
            color[start] = 0

            while queue:
                node = queue.popleft()

                for neighbor in graph[node]:
                    if neighbor not in color:   # Ako neighbor nije obojen, obojimo ga suprotnom bojom od node-a
                        color[neighbor] = 1 - color[node]   # 1 - 1 = 0, 1 - 0 = 1
                        queue.append(neighbor)
                    elif color[neighbor] == color[node]:    # Ako neighbor vec ima istu boju kao node, onda False
                        return False

    return True
```

```python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C'],
    'C': ['B', 'D'],
    'D': ['A', 'C'],
    'E': []
}
```

```python
print(is_bipartite_bfs(graph))
```
